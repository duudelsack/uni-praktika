```python
import numpy as np
from matplotlib import pyplot as pyp
import cv2 as cv
```

```python
image = "/home/farouk/Uni/3. Semester/Bildverarbeitung/Bilder/CMU_Color.png"
new_image = cv.imread(image)
pyp.imshow(new_image)
```

```python
image_blur = cv.GaussianBlur(new_image, (0,0), 5)
pyp.imshow(image_blur)
```

```python
final_image = np.minimum(new_image,  image_blur)
pyp.imshow(final_image)
```

```python
image_blur_1 = np.roll(image_blur, shift=(8,8), axis=(0,1))
pyp.imshow(image_blur_1)
```

```python
final_image_1 = np.minimum(new_image,  image_blur_1)
pyp.imshow(final_image_1)
```
