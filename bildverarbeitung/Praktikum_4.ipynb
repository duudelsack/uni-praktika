```python
import numpy as np
import cv2

#Matrix Implementieren
P = np.array([
    [-524.330, 1075.065, 1876.150, 1652398.600],
    [-308.194, 1728.103,  538.332,  852400.000],
    [ -0.572,    0.614,    0.544,     650.000]
], dtype=np.float64)
```

```python
#Zerlegen mit OpenCV
K, R, C_h, _, _, _, _ = cv2.decomposeProjectionMatrix(P)
```

```python
#K normieren 
K = K / K[2,2]
print(K)
```

```python
#Intirnsische Matrix K
alpha_x = K[0,0]
alpha_y = K[1,1]
s       = K[0,1]
x0      = K[0,2]
y0      = K[1,2]

print(alpha_x, alpha_y, s, x0, y0)
```

```python
#Kamerazentrum C_h
C = (C_h[:3] / C_h[3]).reshape(3)
print("C =", C)
```

```python
#t= -R*C
t = (-R @ C).reshape(3,1)
print("t =", t.ravel())
```

```python
#W=[R∣t] bauen und prüfen K⋅W≈P
W = np.hstack([R, t])
P_rec = K @ W

print("P_rec:\n", P_rec)
```

```python
#skala anpassen
#fehler ausgeben
scale = P[2,3] / P_rec[2,3]   # irgendein Nicht-Null-Element zum Skalieren
P_rec_scaled = P_rec * scale

print("Fehler (Norm):", np.linalg.norm(P - P_rec_scaled))
```
