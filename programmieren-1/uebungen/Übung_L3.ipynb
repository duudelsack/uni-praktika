```python
def oddCount(L):
    if L == []:
        return 0
    else:
        if L[0] % 2 == 0:
            return oddCount(L[1:])
        else:
            return oddCount(L[1:]) + 1
```

```python
def OddSum(L):
    if L == []:
        return 0
    else:
        if L[0] % 2 == 0:
            return OddSum(L[1:])
        else:
            return OddSum(L[1:]) + L[0]
```

```python
def oddsOnly(L):
    if L == []:
        return []
    else:
        if L[0] % 2 != 0:
            return [L[0]] + oddsOnly(L[1:]) 
        else:
            return oddsOnly(L[1:])
```

```python
oddsOnly([1,2,3,4,5,6])
```

```python
def maxOdd(L):
    if L == []:
        return None
    else:
        if L[0] % 2 == 0:
            return maxOdd(L[1:])
        else:
            max_Oddrest = maxOdd(L[1:])
            if max_Oddrest:
                return max(L[0], max_Oddrest)
            else:
                return L[0]
```

```python
def hasConsecutiveDigits(n):
    if n < 10:
        return False
    if (n % 10) == ((n // 10) % 10):
        return True
    return hasConsecutiveDigits(n // 10)
```

```python
def is_palindrome_recursive(strng):
    if len(strng) <= 1:
        return True
    if strng[0] != strng[-1]:
        return False
    return is_palindrome_recursive(strng[1:-1])
```

```python
def num_eights(pos):
    if pos == 0:
        return 0
    count = 1 if pos % 10 == 8 else 0
    
    return count + num_eights(pos // 10)
```

```python
def merge(dict1, dict2):
    merged_dict = dict1.copy()

    for key, value in dict2.items():
        if key in merged_dict:
            merged_dict[key] += value  
        else:
            merged_dict[key] = value  
    return merged_dict
```
