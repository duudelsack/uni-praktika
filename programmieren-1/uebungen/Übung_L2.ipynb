```python
def a_plus_abs_b(a, b):
    """Return a+abs(b), but without calling abs.

    >>> a_plus_abs_b(2, 3)
    5
    >>> a_plus_abs_b(2, -3)
    5
    >>> a_plus_abs_b(-1, 4)
    3
    >>> a_plus_abs_b(-1, -4)
    3
    """
    # YOUR CODE HERE

    if b < 0:
        b = -b
    return a + b

import doctest
doctest.testmod()
```

```python
def two_of_three(i, j, k):
    """Return m*m + n*n, where m and n are the two smallest members of the
    positive numbers i, j, and k.

    >>> two_of_three(1, 2, 3)
    5
    >>> two_of_three(5, 3, 1)
    10
    >>> two_of_three(10, 2, 8)
    68
    >>> two_of_three(5, 5, 5)
    50
    """
    # YOUR CODE HERE

    if i>j and i>k:
        return j*j +k*k
    elif j>i and j>k:
        return i*i + k*k
    else:
        return i*i +j*j
    
import doctest
doctest.testmod()
```

```python
def largest_factor(n):
    """Return the largest factor of n (n > 1 !) that is smaller than n.
    A factor divides a number evenly.

    >>> largest_factor(15) # factors are 1, 3, 5
    5
    >>> largest_factor(80) # factors are 1, 2, 4, 5, 8, 10, 16, 20, 40
    40
    >>> largest_factor(13) # factor is 1 since 13 is prime
    1
    """
    # YOUR CODE HERE

    counter = n-1
    while counter > 0:
        if n % counter == 0:
            return counter
        else:
            counter -= 1

import doctest
doctest.testmod()
```

```python
def sum_digits(y):
    """Sum all the digits of y. y is always nonnegative. Floor division
    and modulo might be helpful.

    >>> sum_digits(10) # 1 + 0 = 1
    1
    >>> sum_digits(4224) # 4 + 2 + 2 + 4 = 12
    12
    >>> sum_digits(1234567890)
    45
    >>> a = sum_digits(123) # use return rather than print in your code!
    >>> a
    6
    """
    # YOUR CODE HERE

    total = 0
    while y > 9:
        total += y % 10
        y = y // 10
    return total + y

import doctest 
doctest.testmod()
```

```python
def getKthDigit(n, k):
    """
    Return the kth digit of n (an integer), starting from 0,
    counting from the right.
    
    >>> getKthDigit(789, 0)
    9
    >>> getKthDigit(789, 1)
    8
    >>> getKthDigit(789, 2)
    7
    >>> getKthDigit(789, 3)
    0
    >>> getKthDigit(-789, 0)
    9
    """
    # YOUR CODE HERE

    if n < 0:
        n = -n
    return (n // 10**k) % 10

import doctest
doctest.testmod()
```

```python
def setKthDigit(n, k, d):
    """
    n is an integer, k is a non-negative integer and d is non-negative
    single digit (0 <= d <= 9). Return the number n with the kth digit
    replaced with d.

    >>> setKthDigit(468, 0, 1)
    461
    >>> setKthDigit(468, 1, 1)
    418
    >>> setKthDigit(468, 2, 1)
    168
    >>> setKthDigit(468, 3, 1)
    1468
    """
    # YOUR CODE HERE

    digit_k = (n // 10**k) % 10
    n -= digit_k * 10**k
    n += d * 10**k
    return n

import doctest
doctest.testmod()
```

```python
def double_eights(n):
    """Return true if n has two eights in a row.
    >>> double_eights(8)
    False
    >>> double_eights(88)
    True
    >>> double_eights(2882)
    True
    >>> double_eights(880088)
    True
    >>> double_eights(12345)
    False
    >>> double_eights(80808080)
    False
    """
    # YOUR CODE HERE

    while n > 10:
        if n % 100 == 88:
            return True
        else:
            n //= 10
    return False
    
import doctest
doctest.testmod()
```

```python
def remove_odd_indices(lst, odd, weight):
    """ 
    lst is a list, odd is a boolean. Return a new list with elements from lst
    removed at certain indices: If odd is True, remove elements at odd indices,
    otherwise remove even indexed elements. The remaining elements are 
    multiplied by weight.
    >>> s = [1, 2, 3, 4]
    >>> t = remove_odd_indices(s, True, 1)
    >>> s
    [1, 2, 3, 4]
    >>> t
    [1, 3]
    >>> l = [5, 6, 7, 8]
    >>> m = remove_odd_indices(l, False, 4)
    >>> m
    [24, 32]
    """
    # YOUR CODE HERE

    new_lst = []
    for (index, elem) in enumerate(lst):
        if odd:
            if index % 2 == 0:
                new_lst.append(elem*weight)
        else:
            if index % 2 != 0:
                new_lst.append(elem*weight)
    return new_lst

import doctest
doctest.testmod()             
```

```python
def merge(lst1, lst2):
    """Merges two sorted lists, lst1 and lst2. The new list contains all
    elements in sorted order.
    >>> merge([1, 3, 5], [2, 4, 6])
    [1, 2, 3, 4, 5, 6]
    >>> merge([], [2, 4, 6])
    [2, 4, 6]
    >>> merge([1, 2, 3], [])
    [1, 2, 3]
    >>> merge([5, 7], [2, 4, 6])
    [2, 4, 5, 6, 7]
    """
    # YOUR CODE HERE

    new_lst = []
    while lst1 != [] and lst2 != []:
        if lst1[0] < lst2[0]:
            new_lst.append(lst1[0])
            lst1 = lst1[1:]
        else:
            new_lst.append(lst2[0])
            lst2 = lst2[1:]
    
    if lst1 == []:
        new_lst += lst2
    if lst2 == []:
        new_lst += lst1
    return new_lst 

import doctest
doctest.testmod()
```

```python
def alternatingSum(lst):
    """
    Return the alternating sum of the integer or float elements
    in the list lst. An alternating sum of a sequence is a sum
    where the sign alternates from positive to negative or
    vice versa.
    >>> alternatingSum([5, 3, 8, 4])
    6
    >>> alternatingSum([])
    0
    """
    # YOUR CODE HERE.

    total = 0
    for (index,elem) in enumerate(lst):
        if index % 2 == 0:
            total += elem
        else:
            total -= elem
    return total

import doctest
doctest.testmod()
```

```python
def is_anagram(s1, s2):
    """
    Decide whether a string s1 and a string s2 are anagrams.
    Use only lists in your solution.
    >>> is_anagram("", "")
    True
    >>> is_anagram("abCdabCd", "abcdabcd")
    True
    >>> is_anagram("abcdaabcd", "aabbcddcb")
    False
    """
    # YOUR CODE HERE

    s1 = s1.lower()
    s2 = s2.lower()

    s1 = list(s1)
    s2 = list(s2)

    s1.sort()
    s2.sort()

    if s1 == s2:
        return True
    else:
        return False
    
import doctest
doctest.testmod()
```

```python
def rotateStringLeft(s, n):
    """
    The function takes a string s and a possibly-negative integer n.
    If n is non-negative, the function returns the string s rotated
    n places to the left. If n is negative, the function returns
    the string s rotated |n| places to the right.
    >>> rotateStringLeft('abcd',  1)
    'bcda'
    >>> rotateStringLeft('abcd', -1)
    'dabc'
    """
    # YOUR CODE HERE

    n = n % len(s)
    return s[n:] + s[:n]
    
import doctest
doctest.testmod()
```

```python
def most_frequent(L):
    """
    Return the most frequent element in a list L, resolving
    ties arbitrarily. L only has elements of scalar types.
    >>> most_frequent([2,5,3,4,6,4,2,4,5])
    4
    >>> most_frequent([2,3,4,3,5,3,6,3,7])
    3
    >>> most_frequent([42])
    42
    >>> most_frequent([])
    None
    """
    # YOUR CODE HERE

    if L == []:
        return print('None')
    newDic = {}
    for elem in L:
        if elem in newDic:
            newDic[elem] += 1
        else:
            newDic[elem] = 1
    return max(newDic, key = newDic.get)

import doctest
doctest.testmod()
```
