```python
def a_plus_b(a, b):
    if b < 0:
        return a + (-b)
    else:
        return a+ b
```

```python
def two_of_three(i, j, k):
    if i > j and i > k:
        return j*j +k*k
    elif j > i and j > k:
        return i*i + k*k
    elif k > i and k > j:
        return i*i + j*j
```

```python
two_of_three(10, 2, 8)
```

```python
def largest_factor(n):
    for counter in range(n-1, 0, -1):
        if n % counter == 0:
            return counter
```

```python
largest_factor(11)
```

```python
def sum_digits(y):
    counter = 0
    while y > 0:
        counter += y % 10
        y //= 10
    return counter
```

```python
sum_digits(4224)
```

```python
def getKthDigit(n, k):
    if n < 0:
        n = -n
    return (n // (10**k)) % 10
```

```python
getKthDigit(12345,3)
```

```python
def setKthdigit(n,k,d):
    if k < 0:
        k = -k
    n -= (((n // (10**k)) % 10)*(10**k))
    n += d*(10**k)
    return n
```

```python
def remove_odd_indices(lst, odd, weight):
    new_lst = []
    for (index, elem) in enumerate(lst):
        if odd:
            if index % 2 == 0:
                new_lst.append(elem*weight)
        else:
            if index % 2 != 0:
                new_lst.append(elem*weight)
    return new_lst

```

```python
s = [5,6,7,8]
remove_odd_indices(s, False, 4)
```

```python
def merge(lst1,lst2):
    new_lst = []
    while lst1 != [] and lst2 != []:
        if lst1[0] < lst2[0]:
            new_lst.append(lst1[0])
            lst1 = lst1[1:]
        else:
            new_lst.append(lst2[0])
            lst2 = lst2[1:]
    new_lst += lst1
    new_lst += lst2
    return new_lst

```

```python
merge([1,3,5,6],[2,3,7,8])
```

```python
def alternatingSum(lst):
    if lst == []:
        return 0
    num = 0
    for (index,elem) in enumerate(lst):
        if index % 2 != 0:
             num += -elem
        else:
            num += elem
    return num
    
```

```python
alternatingSum([5,3,8,4])
```

```python
def rotateStringLeft(s, n):
    """
    The function takes a string s and a possibly-negative integer n.
    If n is non-negative, the function returns the string s rotated
    n places to the left. If n is negative, the function returns
    the string s rotated |n| places to the right.
    >>> rotateStringLeft('abcd',  1)
    'bcda'
    >>> rotateStringLeft('abcd', -1)
    'dabc'
    """
    # YOUR CODE HERE

    s = list(s)
    if n == 0:
            return s
    if n > 0:
        n = n
        L = []
        while n > 0:
            x = s[n-1]
            L.append(x)
            s.remove(x)
            n -= 1
        return s.append(L)



           
import doctest
doctest.testmod()
```

```python
def most_frequent(L):
    """
    Return the most frequent element in a list L, resolving
    ties arbitrarily. L only has elements of scalar types.
    >>> most_frequent([2,5,3,4,6,4,2,4,5])
    4
    >>> most_frequent([2,3,4,3,5,3,6,3,7])
    3
    >>> most_frequent([42])
    42
    >>> most_frequent([])
    0
    """
    # YOUR CODE HERE
    if L == []:
        return 0
    newDic = {}
    for elem in L:
        if elem in newDic:
            newDic[elem] += 1
        else:
            newDic[elem] = 1
    return max(newDic, key = newDic.get)

import doctest
doctest.testmod()
```

```python
def is_anagram(s1, s2):
    """
    Decide whether a string s1 and a string s2 are anagrams.
    Use only dictionaries in your solution.
    >>> is_anagram("", "")
    True
    >>> is_anagram("abCdabCd", "abcdabcd")
    True
    >>> is_anagram("abcdaabcd", "aabbcddcb")
    False
    """
    # YOUR CODE HERE
    s1 = s1.lower()
    s2 = s2.lower()

    s1Dic = {}
    s2Dic = {}

    for elem in s1:
        if elem in s1Dic:
            s1Dic[elem] += 1
        else:
            s1Dic[elem] = 1
    
    for elem in s2:
        if elem in s2Dic:
            s2Dic[elem] += 1
        else:
            s2Dic[elem] = 1

    if s1Dic == s2Dic:
        return True
    else:
        return False

import doctest
doctest.testmod()
```
