```python
def oddCount(L):
    if L == []:
        return 0
    else:
        if L[0]%2 != 0:
            return oddCount(L[1:]) + 1
        else:
            return oddCount(L[1:]) 
```

```python
oddCount([1,2,3,4,5])
```

```python
def oddSum(L):
    if L == []:
        return 0
    else:
        if L[0] % 2 != 0:
            return oddSum(L[1:]) + L[0]
        else:
            return oddSum(L[1:]) 
```

```python
oddSum([1,2,3,4,5])
```

```python

def oddsOnly(L):
    if L == []:
        return []
    else:
        if L[0] % 2 != 0:
            return [L[0]] + oddsOnly(L[1:]) 
        else:
            return oddsOnly(L[1:])
```

```python
oddsOnly([1,2,3,4,5,6,7,8,8,9])
```

```python
def maxOdd(L):
    if not L:
        return None
    else:
        if L[0] % 2 != 0:
            max_Oddrest = maxOdd(L[1:])
            if max_Oddrest:
                return max(L[0], max_Oddrest)
            else:
                return L[0]
        else:
            return maxOdd(L[1:])
    

maxOdd([1,2,3,4,5,6,7,8,8]) 
```
